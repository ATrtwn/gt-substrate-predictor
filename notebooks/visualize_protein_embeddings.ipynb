```python
import pandas as pd
from gt_substrate_predictor.features.protein_emb import *
import torch
```

```python
df = pd.read_csv("../data/UGT.csv")
seqs = df["prot_seq"].tolist()
names = df['UGT_trivial_name'].tolist()
```

```python
emb_pytorch = torch.load("../results/embeddings.pt")
```

```python
emb_numpy = emb_pytorch.cpu().numpy()
```

```python
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
```

```python
pca = PCA(n_components=2)
emb_2d = pca.fit_transform(emb_numpy)  # shape [150, 2]
```

```python
tsne = TSNE(n_components=2, perplexity=30, random_state=42)
emb_2d = tsne.fit_transform(emb_numpy)
```

```python
import plotly.io as pio
pio.renderers.default = "iframe_connected"
```

```python
import plotly.express as px
fig = px.scatter(x=emb_2d[:,0], y=emb_2d[:,1], text=names)
fig.show()
```
